# data acquisition

**propose:** pull spherex fits for a target ra/dec/fov from irsa, or read local fixtures, validate them, and store the raw copy. import data fetching utilities from `src.utils.data_acquisition.py`.

**input:** ra, dec, fov, the data release, and optional irsa credentials from `.env`.

**output:** raw data saved to `data/raw/` with a manifest or checksum log.
